#Assignment : Student Management System
#Description
This system demonstrates the use of OOP concepts like classes,Objects,Instance Variables,Instance methods,Class variables,Class methods and Static methods

```python
class Student:
    university = "Kabarak University"

    def __init__(self, name, student_id, programme):
        self.name = name
        self.student_id = student_id
        self.programme = programme

    def display_details(self):
        print("Name:", self.name)
        print("Student ID:", self.student_id)
        print("Programme:", self.programme)
        print("University:", Student.university)
        print()

    def change_programme(self, new_programme):
        self.programme = new_programme


    @classmethod
    def change_university(cls, new_university):
        cls.university = new_university

    @staticmethod
    def is_valid_id(student_id):
        return len(student_id) >= 5
    
```

##Student Class
The student class stores:
-Name
-Student_Id
-Programme

Methods included:


-display_details()- displays student's name,student Id, programme and university
- change_programme()- Allowa a student to change their programme of study
- change_university()- A class method that changes the university name shared by all student objects
- is_valid_id()- A static method that checks whether a student ID meets the validation rule( at least 5 characters).

```python
student1 = Student("Hamdi", "BDSA1456", "BSc Data Science")
student2 = Student("Amina", "BIT4563", "BSc Information Technology")

student1.display_details()
student2.display_details()
```

#Testing the change_programme() method
 Changing the programme of student 2 from Information Techonology to Finance.

```python
student2.change_programme("BSc Finance")
student2.display_details()
```
